```python
fixed_entries=[
{"question":"what is the annual fee","answer":"The annual fee is Rs 500.","keywords":"fee cost price charge","category":"billing"},
{"question":"how to reset password","answer":"Go to Settings > Reset Password.","keywords":"password reset login","category":"account"},
{"question":"what are your working hours","answer":"We are open 9 AM to 5 PM.","keywords":"hours timing open time","category":"general"},
{"question":"how can i pay the fee","answer":"You can pay via UPI, card, or net banking.","keywords":"pay payment upi fee","category":"billing"}
]
personalized_entries=[
{"question":"how can i contact customer support","answer":"You can contact customer support via email at support@example.com or by calling our helpline.","keywords":"contact support helpline","category":"general"},
{"question":"can i get a refund","answer":"Refunds are processed within 5-7 business days after approval.","keywords":"refund money back","category":"billing"}
]
faq_entries=fixed_entries+personalized_entries
df=pd.DataFrame(faq_entries)
print(df)
```

Q2

```python
def score_entry(query,entry):
    query_words=set(query.lower().split())
    keyword_words=set(entry["keywords"].lower().split())
    question_words=set(entry["question"].lower().split())
    keyword_match=len(query_words.intersection(keyword_words))
    question_match=len(query_words.intersection(question_words))
    return keyword_match*2+question_match

def get_matches(query,df):
    scores=[(idx,score_entry(query,row)) for idx,row in df.iterrows()]
    scores.sort(key=lambda x:x[1],reverse=True)
    return [(df.loc[idx,"question"],df.loc[idx,"answer"],s) for idx,s in scores if s>0]

query="how do i pay my fee"
print(get_matches(query,df))
```

```python
def same_category(category_name,df):
    return df[df["category"]==category_name]

print(same_category("general",df))
```

```python
def get_matches_with_ties(query,df):
    scores=[(idx,score_entry(query,row)) for idx,row in df.iterrows()]
    max_score=max(s for idx,s in scores)
    if max_score==0:
        print("No matching entries found")
        return
    tied_entries=[idx for idx,s in scores if s==max_score]
    if len(tied_entries)>1:
        print("Multiple entries tied with the highest score:")
        for idx in tied_entries:
            print(df.loc[idx,"question"],df.loc[idx,"answer"])
    else:
        idx=tied_entries[0]
        print(df.loc[idx,"question"],df.loc[idx,"answer"])

get_matches_with_ties("fee",df)
get_matches_with_ties("working hours",df)
```
