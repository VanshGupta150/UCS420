Q1

```python
roll_no=1024170180
L=[int(d)*10 for d in str(roll_no)]
print(L)

L.append(100)
print(L)
#appended 100 to the end of L
L.insert(3,5)
print(L)
#inserted 5 at index 3, shifting later elements one position right

L.remove(5)
print(L)
removed=L.pop(10)
print(removed)
print(L)

L_sort=L.copy()
L_sort.sort()
print(L_sort)
L_sort.sort(reverse=True)
print(L_sort)
print(L)
#sorted a copy L_sort instead of L itself, since lists are reference types sorting L directly would also change the order used later for the tuple in Q2

print(f"First three: {L[:3]}")
print(f"Last three: {L[-3:]}")

avg=sum(L)/len(L)
print(f"Avg: {avg}")

new_L=[x for x in L if x>avg]
print(new_L)
```

Q2

```python
tup=tuple(L[:8])
print(tup)
print(f"Max: {max(tup)} Max idx: {tup.index(max(tup))}")
print(f"Min: {min(tup)} Min cnt: {tup.count(min(tup))}")
print(list(tup[::-1]))
#tuples are immutable so there is no in-place reverse, reversing has to build a new list

user_score=int(input("Enter a score: "))
if user_score in tup:
    print(tup.index(user_score))
else:
    print("not present")

try:
    tup[0]=100
except TypeError as e:
    print(e)
    #tuples don't support item assignment because they are immutable, unlike lists

first,second,*rest=tup
print(first,second,rest)
```

Q3

```python
import random
random.seed(roll_no)
numbers=[random.randint(100,900) for _ in range(100)]
print(numbers)

odd_count=len([x for x in numbers if x%2!=0])
print(f"Odd count: {odd_count}")
even_count=len([x for x in numbers if x%2==0])
print(f"Even count: {even_count}")

def is_prime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

prime_count=len([x for x in numbers if is_prime(x)])
primes=[x for x in numbers if is_prime(x)]
print(f"Prime count: {prime_count}")
print(primes)

from collections import Counter
freq=Counter(numbers)
most_common_value,most_common_count=freq.most_common(1)[0]
print(f"Most frequent: {most_common_value} occurs {most_common_count} times")
```

Q4

```python
digits=[int(d) for d in str(roll_no)][:8]
A={d*7 for d in digits}
B={d*9 for d in digits}
print(A)
print(B)
print(f"Union: {A.union(B)}")
print(f"Intersection: {A.intersection(B)}")
print(f"A-B: {A.difference(B)}")
print(f"B-A: {B.difference(A)}")
#difference() gives elements only in one set, symmetric_difference() gives elements in exactly one of the two combined (union of both differences)
print(f"Symmetric difference: {A.symmetric_difference(B)}")
print(f"A subset of B: {A.issubset(B)}")
print(f"B superset of A: {B.issuperset(A)}")

X=int(input("Enter a value to discard from A: "))
A.discard(X)
print(A)
#discard() does nothing if the value is missing while remove() raises a KeyError, so discard() is safer when unsure the value exists
```

```python
my_dict={"name":"Vansh Gupta","roll_no":"1024170180","branch":"COPC","age":21,"city":"Ghaziabad"}
print(my_dict)

my_dict["location"]=my_dict.pop("city")
print(my_dict)

my_dict["cgpa"]=8.5
print(my_dict)

my_dict["age"]=my_dict["age"]+1
print(my_dict)

dict_copy1=dict(my_dict)
popped_value=dict_copy1.pop("branch")
print(dict_copy1)
print(popped_value)
dict_copy2=dict(my_dict)
del dict_copy2["branch"]
print(dict_copy2)
#pop() returns the value of the removed key while del returns nothing

for key,value in my_dict.items():
    print(f"{key} → {value}")

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("email key not found in the dictionary")

friend_dict={"name":"Rahul Sharma","roll_no":"1024170199","branch":"Electronics and Communication Engineering","age":22,"city":"Delhi"}
merged_dict={**my_dict,**friend_dict}
print(merged_dict)
#when both dictionaries share a key, friend_dict's value wins since it is merged in second

string_only_dict={k:v for k,v in my_dict.items() if isinstance(v,str)}
print(string_only_dict)
```
