Q2

```python
arr2=np.array([1,2,3,6,4,5])
print(arr2[::-1])

x=np.array([1,2,3,4,5,1,2,1,1,1])
values,counts=np.unique(x,return_counts=True)
most_frequent_x=values[np.argmax(counts)]
print(most_frequent_x,np.where(x==most_frequent_x)[0])

y=np.array([1,1,1,2,3,4,2,4,3,3])
values_y,counts_y=np.unique(y,return_counts=True)
most_frequent_y=values_y[np.argmax(counts_y)]
print(most_frequent_y,np.where(y==most_frequent_y)[0])
```

Q4

```python
vansh_gupta=np.linspace(10,100,25)
print(vansh_gupta)
print(f"ndim: {vansh_gupta.ndim}")
print(f"shape: {vansh_gupta.shape}")
print(f"size: {vansh_gupta.size}")
print(f"dtype: {vansh_gupta.dtype}")
print(f"nbytes: {vansh_gupta.nbytes}")

vansh_gupta_transpose=vansh_gupta.reshape(25,1)
print(vansh_gupta_transpose)
print(vansh_gupta.T)
#T has no visible effect on a 1D array since transpose only swaps axes and there is only one axis, reshape(25,1) is what actually turns it into a column vector
```

```python
ucs420_vansh_gupta=np.array([10,20,30,40,50,60,70,80,90,15,20,35]).reshape(3,4)
print(ucs420_vansh_gupta)
print(f"mean: {np.mean(ucs420_vansh_gupta)}")
print(f"median: {np.median(ucs420_vansh_gupta)}")
print(f"max: {np.max(ucs420_vansh_gupta)}")
print(f"min: {np.min(ucs420_vansh_gupta)}")
print(f"unique: {np.unique(ucs420_vansh_gupta)}")

reshaped_ucs420_vansh_gupta=ucs420_vansh_gupta.reshape(4,3)
print(reshaped_ucs420_vansh_gupta)

resized_ucs420_vansh_gupta=np.resize(ucs420_vansh_gupta,(2,3))
print(resized_ucs420_vansh_gupta)
```
