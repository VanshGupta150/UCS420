Q2

```python
steps=np.array([
[5000,6200,7100],
[8000,7500,9000],
[4500,5100,4800],
[9000,8500,9500]])
print(f"Total: {np.sum(steps)}")
print(f"Mean: {np.mean(steps)}")
print(f"Max: {np.max(steps)} Min: {np.min(steps)}")
print(f"Per day: {np.sum(steps,axis=0)}")
print(f"Per user: {np.sum(steps,axis=1)}")
print(f"Max position (user,day): {np.unravel_index(np.argmax(steps),steps.shape)}")
```

Q3

```python
original=np.array([1,2,3,4,5,6])
subset=original[1:5]
print(subset)
subset[0]=999
print(original)
print(subset)

subset_copy=original[1:5].copy()
subset_copy[0]=500
print(original)
print(subset_copy)

matrix=np.arange(1,13).reshape(3,4)
print(matrix)
print(matrix[0])
print(matrix[-1])
print(matrix[:,1])
print(matrix[1:3,2:4])

flattened=matrix.flatten()
raveled=matrix.ravel()
print(flattened)
print(raveled)

raveled[0]=100
print(matrix)
print(raveled)
#ravel() returns a view where possible so changing it also changes the original matrix

flattened[0]=200
print(matrix)
print(flattened)
#flatten() always returns a copy so the original matrix is unaffected

print(f"shape: {matrix.shape} ndim: {matrix.ndim} size: {matrix.size} dtype: {matrix.dtype}")
```

Q4

```python
X=np.array([
[6,70,3],
[5,50,6],
[8,80,2],
[4,30,8]
])
y=np.array([40,65,30,85])
print(f"shape: {X.shape} ndim: {X.ndim}")

X_transpose=X.T
print(X_transpose)

XtX=X.T@X
print(XtX)
XtX_inv=np.linalg.inv(XtX)
print(XtX_inv)

beta=XtX_inv@X.T@y
print(beta)

new_user=np.array([5,40,7])
print(new_user@beta)
```
